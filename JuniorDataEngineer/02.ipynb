```python
from pathlib import Path
import pandas as pd


# ====================================
# Configuration
# ====================================

BASE_DIR = Path.cwd()

DATA_DIR = BASE_DIR / "data"

INPUT_FILE = (
    DATA_DIR / "assignment-database.csv"
)

OUTPUT_FILE = (
    DATA_DIR / "02_analysis_results.csv"
)


# ====================================
# Load data
# ====================================

df = pd.read_csv(
    INPUT_FILE
)


# ====================================
# Clean Price column
# ====================================

df["Price"] = pd.to_numeric(
    df["Price"],
    errors="coerce"
)


# ====================================
# 1. Average price for each category
# ====================================

average_price = (
    df.groupby("Category")["Price"]
    .mean()
)


# ====================================
# 2. Average price for Tools
#    Product ID 1030-1040
# ====================================

filtered = df[
    (df["Category"] == "Tools") &
    (df["Product ID"].between(1030, 1040))
]

tools_average = (
    filtered["Price"].mean()
)


# ====================================
# 3. Number of products per category
# ====================================

product_count = (
    df.groupby("Category")
    .size()
)


# ====================================
# 4. Minimum and maximum price
# ====================================

price_range = (
    df.groupby("Category")["Price"]
    .agg(["min", "max"])
)


# ====================================
# Print results
# ====================================

print("Average price by category:")
print(average_price)

print(
    "\nAverage price for Tools "
    "with Product ID 1030-1040:"
)

print(tools_average)

print("\nNumber of products per category:")
print(product_count)

print("\nMinimum and maximum price per category:")
print(price_range)


# ====================================
# Create analysis results
# ====================================

results = []

for category in average_price.index:

    results.append(
        {
            "Category": category,
            "Average Price": average_price[category],
            "Product Count": product_count[category],
            "Minimum Price": price_range.loc[
                category, "min"
            ],
            "Maximum Price": price_range.loc[
                category, "max"
            ]
        }
    )


results_df = pd.DataFrame(
    results
)


# ====================================
# Add Tools-specific result
# ====================================

tools_row = pd.DataFrame(
    [
        {
            "Category": "Tools (ID 1030-1040)",
            "Average Price": tools_average,
            "Product Count": len(filtered),
            "Minimum Price": None,
            "Maximum Price": None
        }
    ]
)


results_df = pd.concat(
    [
        results_df,
        tools_row
    ],
    ignore_index=True
)


# ====================================
# Save analysis results
# ====================================

results_df.to_csv(
    OUTPUT_FILE,
    index=False
)


print(
    f"\nAnalysis results saved to: "
    f"{OUTPUT_FILE}"
)
```

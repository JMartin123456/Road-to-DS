```python
from pathlib import Path

import pandas as pd


# ====================================
# Load
# ====================================

BASE_DIR = Path.cwd()

file_path = BASE_DIR / "data" / "assignment-database.csv"

df = pd.read_csv(file_path)


# ====================================
# Missing values
# ====================================

print("Missing values:")

for row in df[df["Product ID"].isna()].index:
    print(f"A{row + 2}")

for row in df[df["Product Name"].isna()].index:
    print(f"B{row + 2}")

for row in df[df["Category"].isna()].index:
    print(f"C{row + 2}")

for row in df[df["Price"].isna()].index:
    print(f"D{row + 2}")

for row in df[df["Description"].isna()].index:
    print(f"E{row + 2}")


# ====================================
# Product Names starting with space
# ====================================

print("\nProduct Names starting with blank space:")

for row in df[
    df["Product Name"].fillna("").str.startswith(" ")
].index:

    print(f"B{row + 2}")


# ====================================
# Invalid Price
# ====================================

print("\nInvalid Price:")

price = pd.to_numeric(
    df["Price"],
    errors="coerce"
)

for row in df[
    df["Price"].notna() & price.isna()
].index:

    print(
        f"D{row + 2} - {df.loc[row, 'Price']}"
    )


# ====================================
# Duplicate Product ID
# ====================================

print("\nDuplicate Product ID:")

for row in df[df["Product ID"].duplicated(keep=False)].index:

    print(
        f"A{row + 2} - {df.loc[row, 'Product ID']}"
    )


# ====================================
# Other suspicious issues
# ====================================

print("\nOther suspicious issues:")


# Incomplete description

for row in df[
    df["Description"].fillna("").str.strip().str.len() < 10
].index:

    print(
        f"E{row + 2} - "
        "Description may be incomplete"
    )
```
